# Airbnb Data Analysis

This project focuses on data cleaning and exploratory data analysis of an Airbnb dataset containing approximately 100,000 listings.

The analysis uses Python, Pandas, Matplotlib, and Seaborn to clean the data, handle missing values, transform data types, and identify useful patterns in Airbnb listings.

### 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2. Load Dataset

```python
df = pd.read_csv("airbnb.csv")
```

### 3. Initial Data Exploration

### 4. Data Cleaning

### Checking Missing Values

```python
print(df.isnull().sum())
```

```python
df.info()
```

### Handling Missing Values

```python
df["last review"] = pd.to_datetime(df["last review"], errors = 'coerce')
```

```python
df.info()
```

```python
df.fillna({"reviews per month" : 0, "last review" : df['last review'].min()},inplace = True)
```

```python
df.dropna(subset = ['NAME', 'host name'], inplace = True)
```

```python
print(df.isnull().sum())
```

```python
df = df.drop(columns = ['house_rules', 'license'], errors = 'ignore')
```

```python
df.head()
```

```python
#remove dollar sign and convert to float
df['price'] = df['price'].replace('[/$,]', '', regex=True).astype(float)
df['service fee'] = df['service fee'].replace('[/$,]', '', regex=True).astype(float)
```

```python
df.head()
```

### Remove Duplicates

```python
df.drop_duplicates(inplace = True)
```

```python
df.info()
```

### Descriptive Statistics

```python
df.describe()
```

### 5. Exploratory Data Analysis

What is the disribution of the listing prices?

```python
plt.figure(figsize = (10,6))
sns.histplot(df['price'], bins = 50, kde = True, color = 'red')
plt.title('Distribution of Listing Prices')
plt.xlabel('Price $')
plt.ylabel('Frequency')
plt.show()
```

The histogram shows a fairly even distribution of listing prices across different price ranges, indicating no particular concentration of listings in any specific price range. The KDE line helps visualize this even spread more clearly, confirming that the dataset contains listings with a wide variety of prices.

How are different room types distributed?

```python
plt.figure(figsize = (8,5))
sns.countplot(x = 'room type', data = df, color = 'hotpink')
plt.title("Room type distribution")
plt.show()
```

How are the listings distributed across different neighbourhoods?

```python
plt.figure(figsize = (12,8))
sns.countplot(y = 'neighbourhood group', data = df, color = 'lightgreen', order = df['neighbourhood group'].value_counts().index)
plt.title('Number of listings by Neighbourhood Groups')
plt.show()
```

What is the relationship between price and room type

```python
plt.figure(figsize = (10,6))
sns.boxplot(x = 'room type', y = 'price', hue = 'room type', data = df, palette = 'Set1')
plt.title('Price vs Room Type')
plt.legend(title = 'room tyoe')
plt.show()
```

How has the reviews changed with time?

```python
review_over_time = (df.groupby(df['last review'].dt.to_period('M')).size())

plt.figure(figsize = (12,6))
review_over_time.plot(kind = 'line', color = 'red')
plt.title('Number of Reviews Over Time')
plt.xlabel('Year')
plt.ylabel('Average Number of Reviews')
plt.show()
```
